```python
# Data Serialization
import json

data = {"name":"Yaksh","age":23}

json_str = json.dumps(data)
print(json_str)
print(type(json_str))

json_data = json.loads(json_str)
print(json_data)
print(type(json_data))
```

```python
# csv
import csv

with open("example.csv","w",newline='') as f:
    writer = csv.writer(f)
    writer.writerow(["name","age","job"])
    writer.writerow(["Yaksh","23","student"])
    writer.writerow(["Rahul","22","student"])

with open("example.csv","r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
# datetime
import datetime

now = datetime.datetime.now()
print(now)
yesteraday = now - datetime.timedelta(days=1)

print(yesteraday)
print(now.year)
print(now.month)
print(now.day)
print(now.hour)
print(now.minute)
print(now.second)
print(now.microsecond)
```
